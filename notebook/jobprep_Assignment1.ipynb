## Part 7: Pivot Tables

```python
# Part 7: Que 17 --> Create a pivot table: 
""" ● rows = city 
    ● columns = category
    ● values = total amount
"""
Final_Data['amount'] = Final_Data['amount'].fillna(0)
pivot_table = pd.pivot_table(
    Final_Data,
    values='amount',
    index='city',
    columns='category',
    aggfunc='sum',
    fill_value=0
)

print(pivot_table)
```

## Part 8: Melt

```python
# Part 8: Que 18 --> Convert the pivot table back into long format using melt()
pivot_table_reset = pivot_table.reset_index()
melted_data = pd.melt(
    pivot_table_reset,
    id_vars='city',          # column to keep fixed
    var_name='category',     # column name for melted columns
    value_name='total_amount'  # values column
)

print(melted_data)
```

## Part 9: Final Challenge (Very Important)

```python
# Part 9: Que 19 --> Final Challenge (Very Important) 19. Write a cleaning pipeline : 
# ● remove duplicates 
# ● handle missing values
# ● merge datasets

import pandas as pd

def clean_and_merge(Customers, Orders, Products):
    
    # Convert to DataFrames
    Customers = pd.DataFrame(Customers)
    Orders = pd.DataFrame(Orders)
    Products = pd.DataFrame(Products)
    
    # -----------------------------
    # 1. Remove Duplicates
    # -----------------------------
    Customers = Customers.drop_duplicates(subset='customer_id')
    Orders = Orders.drop_duplicates(subset='order_id')
    Products = Products.drop_duplicates(subset='product')
    
    # -----------------------------
    # 2. Handle Missing Values
    # -----------------------------
    # Fill missing amount with 0
    Orders['amount'] = Orders['amount'].fillna(0)
    
    # Fill missing age with mean
    if 'age' in Customers.columns:
        Customers['age'] = Customers['age'].fillna(Customers['age'].mean())
    
    # Fill missing city/name with 'Unknown'
    Customers['city'] = Customers['city'].fillna('Unknown')
    Customers['name'] = Customers['name'].fillna('Unknown')
    
    # -----------------------------
    # 3. Merge Datasets
    # -----------------------------
    merged_data = pd.merge(Orders, Customers, on='customer_id', how='left')
    final_data = pd.merge(merged_data, Products, on='product', how='left')
    
    return final_data


# Run pipeline
Final_Data = clean_and_merge(Customers, Orders, Products)

print(Final_Data)
print("\nMissing values after cleaning:\n", Final_Data.isnull().sum())
```
